MOHMMED AQEEB PATHAN — 242A003


```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
import nltk
from nltk import pos_tag, word_tokenize
from nltk.chunk import RegexpParser

nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
text1 = "Penny visits her brother Thomas at his bakery."
text2 = "The bakery is lively and people of all backgrounds are there."
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
tokens = word_tokenize(text2)
tagged = pos_tag(tokens)
print(tagged)
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
grammar = "NP: {<DT>?<JJ>*<NN>}"
chunk_parser = RegexpParser(grammar)
result = chunk_parser.parse(tagged)
print(result)
```
