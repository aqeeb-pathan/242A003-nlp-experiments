```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
# Import Libraries & Download Resources
import string
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
import re
import emoji

nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
# Original Text
text = '''Hey everyone!!! 👋

I'm REALLY excited 😊 because I've just completed my first Machine Learning project on 23rd July, 2026. The project analyzed over 12,500 customer reviews collected from https://www.example.com and another dataset downloaded from www.kaggle.com.

Unfortunately, some reviews weren't useful—they contained only words like "the", "is", "and", "of", "to", etc. Others had strange symbols such as @#$%^&*, repeated punctuation!!!!!!, HTML tags like <br>, and emojis 😂😂😂.

You can contact me at vijay.shanker@eng.rizvi.edu.in or support123@company.co.in if you'd like the dataset. Follow us on Twitter @AI_ResearchLab and use hashtags like #MachineLearning #NLP #DataScience.

The model's accuracy increased from 82.5% to 91.8%, while the training time decreased by nearly 35%. It processed 1,25,000 records in less than 2.5 hours and consumed around 3.7 GB of RAM.

Students were studying, analyzing, processing, predicting, running, writing, and comparing different algorithms. The researchers discovered better techniques after performing extensive experiments and evaluating multiple models.

Don't worry if the results aren't perfect—you're learning! Let's improve them together. I'll upload the final report by 30/07/2026.

Thanks!!!'''
text
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
# Text to Lowercase
text_lower = text.lower()
text_lower
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")

import re

# Remove URLs
processed_text = re.sub(r'http\S+|www\.\S+', '', text_lower)

# Remove Twitter mentions and hashtags
processed_text = re.sub(r'@\s*\w+', '', processed_text)
processed_text = re.sub(r'#\s*\w+', '', processed_text)

# Remove numbers and any immediately attached non-whitespace characters
processed_text = re.sub(r'\d+\S*', '', processed_text)

# Normalize spaces after all removals
text_no_url = re.sub(r'\s+', ' ', processed_text).strip()

text_no_url
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
# Removing all emojis
text_no_emoji = emoji.replace_emoji(text_no_url, replace='')
text_no_emoji
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
punctuations = list(string.punctuation)
print(punctuations)
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
# Tokenized Text
tokenized_text = word_tokenize(text_no_emoji)
print(tokenized_text)
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
# Removing Punctuation
cleaned_text1 = []
for word in tokenized_text:
  if word not in punctuations:
    cleaned_text1.append(word)
print(cleaned_text1)
print(len(cleaned_text1))
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
print(stopwords.words('english'))
print(len(stopwords.words('english')))
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
# Removing Stopwords
cleaned_text2 = []
for word in cleaned_text1:
  if word not in stopwords.words('english'):
    cleaned_text2.append(word)
print(cleaned_text2)
print(len(cleaned_text2))
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
# Stemming
stemmer = PorterStemmer()
stemmed_text = []
for word in cleaned_text2:
  stemmed_text.append(stemmer.stem(word))
print(stemmed_text)
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_text = []
for word in cleaned_text2:
  lemmatized_text.append(lemmatizer.lemmatize(word))
print(lemmatized_text)
```

```python
print("MOHMMED AQEEB PATHAN - UIN: 242A003")
print("Original Text:")
print(text)
print("\nStemmed Text:")
print(stemmed_text)
print("Doug was here...")
print("\nLemmatized Text:")
print(lemmatized_text)
```
